```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    "../data/restricted/kidney_waitlist_analytic.csv.gz",
    dtype={
        "PREV_TX": "string",
        "TX_DATE": "string",
        "DON_TY": "string",
        "MULTIORG": "string",
        "ORGAN": "string",
        "TRR_ID_CODE": "string",
    },
    low_memory=False,
)
```

```python
num_df = df.select_dtypes(include=["int64", "float64"])
num_df
```

```python
corr_matrix = numeric_df.corr()[["event adverse"]]

sns.heatmap(corr_matrix, annot=True)

plt.title("Correlation with Event Adverse")
plt.show()
```
