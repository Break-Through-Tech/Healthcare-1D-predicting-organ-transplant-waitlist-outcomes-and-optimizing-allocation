```python
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("../data/processed/task3_encoded_unscaled.csv")
df.shape
```

```python
# column groups, copied over from encoding-scaling
target_cols = ["outcome", "event_adverse", "event_transplant", "censored", "days_to_event"]

leaky = ["DIALYSIS_AFTER_LISTING"] #may be known after listing

numerical_cols = [
    "BMI_TCR", "INIT_CPRA", "INIT_AGE", "functional_percent",
    "DIALYSIS_DURATION", "DIALYSIS_DURATION_LOG", "AGE_DIALYSIS_INTERACTION",
]

feature_cols = [c for c in df.columns if c not in target_cols]
# only scale numeric columns that are still in feature set
numerical_cols = [c for c in numerical_cols if c in feature_cols]
print(len(feature_cols), "feature columns")
```

```python
# split
# split by 20%-- this gives us lots of rows to examine, a higher split would be too much
X = df[feature_cols]

#pulls out the answer columns
y = df[["outcome", "event_adverse", "event_transplant", "censored", "days_to_event"]]

# stratify y means to split within each category seperately then combine the pieces
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y["outcome"], random_state=42
)
print(X_train.shape, X_test.shape)

# check that split preserved rates
print("Train rates:")
print(y_train[["event_adverse", "event_transplant", "censored"]].mean())

print("Test rates:")
print(y_test[["event_adverse", "event_transplant", "censored"]].mean())
```

```python
# scale, but fit on train only

scaler = StandardScaler()
X_train = X_train.copy()
X_test = X_test.copy
X_train[numerical_cols] = scaler.fit_transform(X_train[numerical_cols])
X_test[numerical_cols] = scaler.transform(X_test[numerical_cols])
```

```python
# Save everything
X_train.to_csv("../data/processed/X_train.csv", index=False)
X_test.to_csv("../data/processed/X_test.csv", index=False)
y_train.to_csv("../data/processed/y_train.csv", index=False)
y_test.to_csv("../data/processed/y_test.csv", index=False)
```
